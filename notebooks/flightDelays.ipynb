### loading the Datasets using pandas

```python
airport_raw = pd.read_csv("../data/Raw/airports.csv")
airport_raw.head()
```

```python
delays_raw = pd.read_excel("../data/Raw/RP4_APT_ATFM_ARR_2026_Jan_Feb.xlsx", sheet_name="APT_ATFM_APT", header=4)
delays_raw.head()
```

```python
## Phase 2: Data Filtering and Merging

### Step 1: Create Country Mapping and Filter Airports
```

```python
# Create country-to-ISO code mapping
country_iso_mapping = {
    'Germany': 'DE',
    'Spain': 'ES',
    'Italy': 'IT',
    'Portugal': 'PT',
    'Netherlands': 'NL',
}

# Extract the ISO codes for filtering
iso_codes = list(country_iso_mapping.values())
countries_list = list(country_iso_mapping.keys())

print(f"Selected ISO Codes: {iso_codes}")
print(f"Selected Countries: {countries_list}")

# Filter airport_raw for the selected countries
airports_europe = airport_raw[airport_raw['iso_country'].isin(iso_codes)].copy()
print(f"\nAirports filtered: {len(airports_europe)} records")
print(f"Countries in airport data: {airports_europe['iso_country'].unique()}")
airports_europe.head()
```

### Step 2: Filter Delays Data for Selected Countries

```python
# Filter delays_raw for the selected countries
delays_europe = delays_raw[delays_raw['State'].isin(countries_list)].copy()
print(f"Delays filtered: {len(delays_europe)} records")
print(f"Countries in delays data: {delays_europe['State'].unique()}")
print(f"\nICAO codes in delays data:\n{delays_europe['ICAO'].unique()}")
delays_europe.head()
```

### Step 3: Merge the DataFrames on ICAO Code

```python
# Perform outer merge on ICAO code
# airports_europe will provide airport data (ICAO is from icao_code column)
# delays_europe will provide delay metrics (ICAO is already named ICAO)

merged_europe = pd.merge(
    airports_europe,
    delays_europe,
    left_on='icao_code',  # ICAO code in airports dataset
    right_on='ICAO',       # ICAO code in delays dataset
    how='outer'            # Outer join - keeps all records from both
)

print(f"Merged dataframe shape: {merged_europe.shape}")
print(f"\nNumber of airports matched: {len(merged_europe)}")
print(f"Countries in merged data: {merged_europe['iso_country'].unique()}")
print(f"\nColumns in merged dataframe:\n{merged_europe.columns.tolist()}")
merged_europe.head()
```

### Step 4: Validate the Merge

```python
# Validation checks
print("=" * 60)
print("MERGE VALIDATION REPORT")
print("=" * 60)

# Check 1: Countries represented
print("\n1. Countries represented in merged data:")
for country in merged_europe['State'].unique():
    count = len(merged_europe[merged_europe['State'] == country])
    print(f"   {country}: {count} airports")

# Check 2: Data completeness
print(f"\n2. Data completeness check:")
critical_cols = ['name', 'State', 'ICAO', 'Apt. ATFM dly [min./arr.]', 'FLTS [ARR]']
for col in critical_cols:
    missing = merged_europe[col].isna().sum()
    print(f"   {col}: {missing} missing values")

# Check 3: Summary statistics
print(f"\n3. Delay statistics:")
print(f"   Average ATFM delay (min): {merged_europe['Apt. ATFM dly [min./arr.]'].mean():.2f}")
print(f"   Max ATFM delay (min): {merged_europe['Apt. ATFM dly [min./arr.]'].max():.2f}")
print(f"   Total flights (arrivals): {merged_europe['FLTS [ARR]'].sum():.0f}")

print(f"\n4. Final merged dataset:")
print(f"   Shape: {merged_europe.shape}")
print(f"   Total unique airports: {merged_europe['name'].nunique()}")
print("\n✓ Merge completed successfully!")
print("=" * 60)
```
